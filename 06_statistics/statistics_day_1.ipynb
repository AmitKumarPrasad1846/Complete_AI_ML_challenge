# What is statistics ?

### Statistics is branch of mathematics which deals with collecting,understanding,finding patter and making decision

- it help to understand data and find meaningful information from it
- it helps to convert raw data into useful information
- data -> understanding -> decision

**`eg -> Netflix want to know what type of movies users like`**

* It can collect data
- which movie users watch
- how long they watch
- which movies they stopped watching
- rating giving by users

# Why we need statistics ?
- To understand the pattern in the data
- To make decisions
- To summarize huge data
- Better decisions
- To test Ideas

# What statistics do ?
 - It mainly does the following things :- 
    - 1. **collect data** -> gather data`(surveys,observation,experiments)`
    - 2. **organize data** -> Arrange data properly
    - 3. **summarize** -> make data simple - **(min,max,average)**
    - 4. **analyse** -> find `pattern`,`trends`,`relationship` and `predictions`
    - 5. **preict** -> make decision

# Types of statistics

- There are two types of statistics

    1. **Descriptive statistic** -> Describe the data

        - It summarize the existing data
        - Used to analyze and present the data into meaning way
            - Ex -> `mean`,`median`,`mode`,`variance`,`std`

    2. **Inferential statistics** -> make prediction or conclusion

*draw conclusion or make prediction about a population based based on smaple
*example ->test,hypothises,(z,t)


POLULATION VS SAMPLE

*POLULATION -> entire group or superset
*SAMPLE -> small part/small group from the large data or subset of population
