```python
'''
Quartile -> from the quarter means 4 equal part

*it divides/split the data into 4 eqaul part
three main quariles
1.Q1(quartile 1) -> 25th percentile of the data
2.Q2(quartile 2) -> 50th percentile of the data
3.Q3(quartile 3) -> 75th percentile of the data

'''
data = [12, 15, 18, 20, 22, 25, 30, 35, 40]
#what is the percentage of data lies approximately below Q1

q1=np.percentile(data,25) #25% of data lies below 18
print(q1)

#what is the percentage of data lies below Q3
q2=np.percentile(data,75) #75% of data lies below 30
print(q2)
```

```python
'''
Outliers -> a value that is unusually far away from the most of the other values

'''
import seaborn as sns
salary=[10000,20000,30000,40000,50000,100000,-100000]
print(np.mean(salary))
sns.boxplot(salary)
```

# Mean is very sensitive for outliers
# Median is not that much sensitive for outliers

```python
'''
is the outliers are bad for my data ?

how can we remove the outlier ?

using one method called IQR(inter quartile range )

IQR=Q3-Q1

lower bound -> the smallest value allowed in the range -> q1-1.5*IQR
upper bound -> the largest value allowed in the range -> q3+1.5*IQR

(any value<lower_bound) and (any value>upper_bound) ----> outliers


'''
```

```python
data3=[-34,50,55,64,58,62,69,70,73,56,80,92,390]

q1=np.percentile(data3,25)
q3=np.percentile(data3,75)

IQR=q3=q1
print(IQR)

sns.boxplot(data3)
```

```python
lower_bound=q1-1.5*IQR
upper_bound=q3+1.5*IQR
```

```python
clean_data=[]

for value in data3:
  if value<lower_bound or value>upper_bound:
    print('outlier',value)
  else:
    clean_data.append(value)

print(clean_data)
```
