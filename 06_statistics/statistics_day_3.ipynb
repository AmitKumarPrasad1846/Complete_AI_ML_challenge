```python
'''
Quartile -> from the quarter means 4 equal part

*it divides/split the data into 4 eqaul part
three main quariles
1.Q1(quartile 1) -> 25th percentile of the data
2.Q2(quartile 2) -> 50th percentile of the data
3.Q3(quartile 3) -> 75th percentile of the data

'''
data = [12, 15, 18, 20, 22, 25, 30, 35, 40]
#what is the percentage of data lies approximately below Q1

q1=np.percentile(data,25) #25% of data lies below 18
print(q1)

#what is the percentage of data lies below Q3
q2=np.percentile(data,75) #75% of data lies below 30
print(q2)
```
