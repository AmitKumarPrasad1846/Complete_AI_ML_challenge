# Create a class Student with private attributes name and marks. Provide getter and setter methods. Prevent marks from being set below 0 or above 100. 

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.set_marks(marks)

    # Getter and Setter for Name
    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    # Getter and Setter for Marks
    def get_marks(self):
        return self.marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.marks = marks
        else:
            print("Invalid marks! It must be between 0 and 100.")

# running the class...

s = Student("Amit", 85)
print(s.get_name(), s.get_marks())

s.set_marks(105) # output : Invalid marks! It must be between 0 and 100.
```

# Print first N prime numbers efficiently.

```python
n = int(input("How many Prime no. you want to print? = "))

prime = []
num = 2

while len(prime) < n:
    for i in range(2, num):
        if num % i == 0:
            break
    else:
        prime.append(num)
    num += 1

print(prime)
```
