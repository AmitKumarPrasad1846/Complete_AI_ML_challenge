## 1. Introduction to Pandas
Pandas is a Python library used for **data manipulation** and **data analysis**. It helps to load, clean, transform, analyze, and explore data.

* **Origin:** Created by Wes McKinney in 2008.
* **Name Origin:** Stands for **"Panel Data System"**.
* **Purpose:** Built to solve real-world data issues (handling missing values, duplicates, column labels) beyond simple math operations.

### Data Lifecycle / Pipeline
1. **Load:** Import data into Python (CSV, Excel, SQL, etc.).
2. **Clean:** Remove/fill missing data and fix bad formatting.
3. **Manipulate:** Filter, sort, and merge datasets.
4. **Explore:** Discover patterns and insights (EDA).
5. **Analyze:** Feed clean data into Machine Learning (ML) pipelines.

---

## 2. Core Data Structures

Pandas operates on two main data structures:

| Data Structure | Dimensions | Description | Concept |
| :--- | :--- | :--- | :--- |
| **Series** | 1D | One-dimensional labeled array | A single column of data with an index |
| **DataFrame** | 2D | Two-dimensional tabular data | A table with rows and columns (Multiple Series) |

---

## 3. Pandas Series in Detail
* **Definition:** A 1D labeled array capable of holding mixed data types (numbers, strings, booleans).
* **Formula:** `Series = Data + Index`

* **Dictionary Values** ➡️ Become the Series **Data**.
