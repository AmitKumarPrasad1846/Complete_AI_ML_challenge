# SEABORN

- It is also used for the visualization of the data in graph or chart
    - It is python libray built on top of matplotlib
    - It helps you to draw beautiful and informative plots easily
    - It is good statistical and categorical data

### Advantage :
- less code
- more beautiful visualization

## library --> `seaborn library`

## Types of Plot :-
- line plot
- bar plot
- heat map
- count plot
- violin plot

```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame(
    {
        'days': ['mon', 'tue', 'wed', 'thur', 'fri'],
        'sales': [200, 250, 310, 480, 590],
        'visitors': [50, 60, 40, 70, 75],
        'gender': ['m', 'f', 'm', 'f', 'm'],
    }
)
print(df)
```

```python
# line plot ->
# how to plot -> sns.lineplot(x=data, y=data, data=data) (x_axis, y_axis, dataframe_name)

sns.lineplot(
    x='days',
    y='sales',
    data=df,
    marker='o',
    color='red',
    markersize=20,
)
plt.title('days vs sales')
plt.show()
```

```python
# hue --> this create multipleline based on category(it groups the data based on category)

sns.lineplot(
    x='days',
    y='sales',
    data=df,
    marker='o',
    color='red',
    markersize=20,
    hue='gender',
)
plt.title('days vs sales')
plt.show()
```

```python
# Bar plot

sns.barplot(x='days',y='sales',data=df,color='green',edgecolor='black')
plt.title('bar graph using seaborn')
plt.show()
```

```python
# hue --> this create multipleline based on category(it groups the data based on category)

sns.lineplot(x='days',y='sales',data=df,marker='o',color='red',markersize=20,hue='gender')
plt.title('days vs sales')
plt.show()
```

```python
# Histogram --> attached bar like graph
# how to plot --> sns.histplot()

sns.histplot(df['sales'],color='cyan')
plt.title('histogram using seaborn')
plt.show()
```

```python
#KDE -> kernal density estimation
#it will give smooth curve that show how data is distributed

sns.histplot(df['sales'],color='red',edgecolor='black',kde=True)
plt.title('histogram using seaborn')
plt.show()
```

```python
# scatter plot
# how to plot -> sns.scatterplot(x,y,data)

df
```

```python
sns.scatterplot(x='visitors',y='sales',data=df,color='red',hue='gender')
plt.title('visitors vs sales')
plt.grid()
plt.show()
```

```python
'''
box plot

how to plot --> sns.boxplot(data)

'''
sns.boxplot(df['sales'])
plt.show()
```

```python
'''
violin plot

how to plot --> sns.violinplot(data)
'''
sns.violinplot(df['sales'])
plt.show()
```

```python
'''
COUNT PLOT

it is used to count the number of occurrence of each catogory in categorical variable
*how to plot --> sns.countplot(x,data)

'''
sns.countplot(x='gender',data=df,color='yellow',edgecolor='black')
plt.title('count plot')
plt.show()
```
