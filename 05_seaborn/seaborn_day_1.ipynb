# SEABORN

- It is also used for the visualization of the data in graph or chart
    - It is python libray built on top of matplotlib
    - It helps you to draw beautiful and informative plots easily
    - It is good statistical and categorical data

### Advantage :
- less code
- more beautiful visualization

## library --> `seaborn library`

## Types of Plot :-
- line plot
- bar plot
- heat map
- count plot
- violin plot

```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame(
    {
        'days': ['mon', 'tue', 'wed', 'thur', 'fri'],
        'sales': [200, 250, 310, 480, 590],
        'visitors': [50, 60, 40, 70, 75],
        'gender': ['m', 'f', 'm', 'f', 'm'],
    }
)
print(df)
```

```python
# line plot ->
# how to plot -> sns.lineplot(x=data, y=data, data=data) (x_axis, y_axis, dataframe_name)

sns.lineplot(
    x='days',
    y='sales',
    data=df,
    marker='o',
    color='red',
    markersize=20,
)
plt.title('days vs sales')
plt.show()
```

```python
# hue --> this create multipleline based on category(it groups the data based on category)

sns.lineplot(
    x='days',
    y='sales',
    data=df,
    marker='o',
    color='red',
    markersize=20,
    hue='gender',
)
plt.title('days vs sales')
plt.show()
```

```python
# Bar plot

sns.barplot(x='days',y='sales',data=df,color='green',edgecolor='black')
plt.title('bar graph using seaborn')
plt.show()
```

```python
# hue --> this create multipleline based on category(it groups the data based on category)

sns.lineplot(x='days',y='sales',data=df,marker='o',color='red',markersize=20,hue='gender')
plt.title('days vs sales')
plt.show()
```

```python
# Histogram --> attached bar like graph
# how to plot --> sns.histplot()

sns.histplot(df['sales'],color='cyan')
plt.title('histogram using seaborn')
plt.show()
```

```python
#KDE -> kernal density estimation
#it will give smooth curve that show how data is distributed

sns.histplot(df['sales'],color='red',edgecolor='black',kde=True)
plt.title('histogram using seaborn')
plt.show()
```

```python
# scatter plot
# how to plot -> sns.scatterplot(x,y,data)

df
```

```python
sns.scatterplot(x='visitors',y='sales',data=df,color='red',hue='gender')
plt.title('visitors vs sales')
plt.grid()
plt.show()
```

```python
'''
box plot

how to plot --> sns.boxplot(data)

'''
sns.boxplot(df['sales'])
plt.show()
```

```python
'''
violin plot

how to plot --> sns.violinplot(data)
'''
sns.violinplot(df['sales'])
plt.show()
```

```python
'''
COUNT PLOT

it is used to count the number of occurrence of each catogory in categorical variable
*how to plot --> sns.countplot(x,data)

'''
sns.countplot(x='gender',data=df,color='yellow',edgecolor='black')
plt.title('count plot')
plt.show()
```

```python
sns.countplot(x='gender',data=df,edgecolor='black',palette='Set2')
plt.title('count plot')
plt.show()
```

```python
'''
HEATMAP ---> it is a table of number displayed in color(matrix)

*darker or brighter color represent high or lower values
*how to plot --> sns.heatmap()

co-relation --> the relationship between 2 coloum or variable (-1 to 1)
1.positive
2.negative
3.zeor co-relation

'''
```

```python
data=[[10,20,30],[40,50,60],[70,80,90]]
sns.heatmap(data)
plt.show()
```

```python
sns.heatmap(df[['sales','visitors']].corr()) #corr() -> to get the co-relation between columns
plt.show()
```

```python
#annot=True --> it will show the co-relation avlues on heatmap
sns.heatmap(df[['sales','visitors']].corr(),annot=True)
plt.show()
```

```python
#cmap -> to change the heatmap color
sns.heatmap(df[['sales','visitors']].corr(),annot=True,cmap='coolwarm')
plt.show()
```

```python
'''
pair plot --->

*it is graph that show relationship between every numerical columns in dataset
*it will create scatter and histogram
*histogram is in diagonal
*how to plot --> sns.pairplot(data)
'''
df
```

```python
'''
area plot

it is type of plot which show. the area using filled color (lineplot + filled color)
*it is just like a lineplot but the area below the lline will be filled with color,
*instead of just showing line it show the volume using filled color

'''
```

```python
df2=pd.DataFrame({'days':[1,2,3,4,5,6,7],
                  'sales':[100,200,150,230,460,280,560]})
df2
```

```python
sns.lineplot(x='days',y='sales',data=df2,color='red')
plt.show()
```

```python
plt.fill_between(df2['days'],df2['sales'],color='red')
plt.title('area plot')
plt.show()
```

```python
#alpha --> controls the transparency of color
plt.fill_between(df2['days'],df2['sales'],color='red',alpha=0.3)
plt.title('area plot')
plt.show()
```
