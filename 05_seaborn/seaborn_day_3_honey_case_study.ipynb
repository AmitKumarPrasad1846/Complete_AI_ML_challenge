1.whcih state is having highest honey production throughout all the year?Represent using bar graph ?

```python
high_prod=df.groupby('state')['production'].sum().sort_values(ascending=False).reset_index()
high_prod
```

```python
plt.figure(figsize=(10,8))
sns.barplot(data=high_prod,x='production',y='state',palette='hsv')
plt.show()
```

2. Which state is having highest and lowest yield per colony? Represent using bar graph

```python
high_yield_col=df.groupby('state')['yield_per_colony'].mean().sort_values(ascending=False).reset_index()
high_yield_col
```

```python
plt.figure(figsize=(10,8))
sns.barplot(data=high_yield_col,x='yield_per_colony',y='state',palette='hsv')
plt.show()
```

3.which year witnessess the highest profit in honey production ?use bar plot to represent this ?


```python
profit_by_year = df.groupby('year')['value_of_production'].sum().sort_values(ascending=False).reset_index()
display(profit_by_year.head())
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(x='year', y='value_of_production', data=profit_by_year, palette='viridis')
plt.title('Total Value of Honey Production by Year')
plt.xlabel('Year')
plt.ylabel('Value of Production')
plt.xticks(rotation=45)
plt.show()
```
