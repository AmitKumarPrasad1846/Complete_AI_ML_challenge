# Probability

## **Probability**
#### It is a branch of mathematics that measures the chances of an event occuring

- Possibility of occuring an event
- Prob = no. of favourable/total number of outcomes

# Why we need Probability ?
- To measure the uncertainity
- To make predictions
- To make better decisions
- To handle risk

```python
'''
key points/terms

1.Experiment -> any process or action that gives an uncertain result
ex->tossing a coin,rolling dice

2.Outcome -> An outcome os one possible result of an experiment
ex-> tossing c coin(experiment) ------> head(one out come) and also tail(another outcome )

3.Sample space -> set/collections of all possible outcomes (S)
ex -> rooling a dice -> S ={1,2,3,4,5,6}

4. Event -> An event is one or more outcomes that we are intrested in...
ex -> getting head,getting 6 while rolling dice
'''
```

```python
'''
Type of events

1.simple event -> An event has only one possible outcome
ex- rolling a dice -> S={1,2,3,4,5,6} but result will always be only one value

2.Compound event -> An event has more than one possible outcomes
ex- rolling a dice -> S={1,2,3,4,5,6}
prob of getting an even number -> {2,4,6} - 3 outcomes
'''
```

```python
'''
1.Marginal probability -> probability of one event happening idependently

*focusing on single event not considering any other event

ex - chocolates in box
total = 50 chocolates ------->20 milk chocolates ,30 dark chocolates

Question -> probability of dark chocolates -->
'''
```

```python
#question -> A bag has 8 black ball and 12 blue balls ,Find the probability of blue balls?

black_balls=8
blue_balls=12
total=black_balls+blue_balls
prob=blue_balls/total   #12/20 = 0.6
print('The marginal probability is =',prob)
```
