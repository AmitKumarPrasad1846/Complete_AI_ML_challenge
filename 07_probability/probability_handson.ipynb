```python
'''
1.Student performance analysis

genrate 40 random marks between 35 to 100 using numpy

*mean
*median
*mode
*variance
*standard deviation (all using stats)

then determine whether the data has high or low variation based on the standard deviation

'''
```

```python
import numpy as np
from scipy import stats

marks=np.random.randint(35,101,40)

mean1=np.mean(marks)
median1=np.median(marks)
mode1=stats.mode(marks)
var1=np.var(marks)
std1=np.std(marks)


print(mean1)
print(mode1)
print(median1)
print(var1)
print(std1)
```

```python
'''
2.temp=[23,24,22,25,23,24,26,25,24,23,22,24,25,26,80,23,24,25,22,24]

using IQR method
*calculate Q1 and Q3
*calculate IQR
*calculate lower and upper bounds
*identify outliers

'''
```

```python
temp=[23,24,22,25,23,24,26,25,24,23,22,24,25,26,80,23,24,25,22,24]

Q1 = np.percentile(temp,25)
print("Q1 = ", Q1)
Q3 = np.percentile(temp,75)
print("Q3 = ", Q3)

IQR = Q3-Q1
print("IQR = ",IQR)

lower_bound = Q1 - 1.5 * IQR
print("lower_bound = ",lower_bound)

upper_bound = Q3 + 1.5 * IQR
print("upper_bound = ",upper_bound)

outlier = [x for x in temp if x < lower_bound or x > upper_bound]
print("Outlier =", outlier)
```
