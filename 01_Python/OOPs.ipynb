**Inheritance:** Property in which a child class inherit(can access) the things(Properties and methods) of Parent class


We derive the properties of parent class in child class



```python
class A:
  def functionA(self):
    print("function from A")

class B:
  def functionB(self):
    print("function from B")

obj=B()
obj.functionA()   # i cant access the another class data
```

```python
class A:
  def a(self):
    print("function from A")

#syntax= class Child(Parent):
class B(A):   #inheritance B inherits the property of A
  def b(self):
    print("function from B")

obj=B()
obj.a()

# Note: Child will be able to access the property of Parent but Parent cant access the properties of child
```

**Types of Inheritance:**
- Single inheritance
- multiple inheritance
- hierarchical inheritance
- Multi level inheritance

1. **Single Inheritance:** one Parent, one child

image.png

```python
class Animal:   #Parent
  def sleep(self):
    print("Animals can sleep")

# Single Inheritance
class Dog(Animal):  #Child
  def Sound(self):
    print("Dog can Bark")

d=Dog()
d.Sound()
d.sleep()
```

**2. Multiple Inheritance:** Multiple Parents, Single child

image.png

```python
class Mother:  #parent1
  def fun(self):
    print("Behaviour is Polite")

class Father:  #parent2
  def fun1(self):
    print("Behaviour is Strict")

#syntax: class Child(Parent1,Parent2):
class Son(Mother,Father): #Child
  def fun2(self):
    print("Behaviour is friendly")

obj=Son()
obj.fun1()
obj.fun()
obj.fun2()
```

**3. Hierarchical inheritance:** Single Parent with Multiple Childs

image.png

```python
class Animal:   #Parent
  def breath(self):
    print("Animal can Breath")

  def eat(self):
    print("Animal can Eat")

class Dog(Animal):  # Child1
  def sound(self):
    print("Dogs can Bark")

class Cat(Animal):  # Child2
  def sound(self):
    print("Cat can Meow")


obj=Dog()
obj.sound()
obj.eat()

print()

obj1=Cat()
obj1.sound()
obj1.breath()
```

4. multi level Inheritance: A -> B -> C, C is the child of B, and B is the Child of A

Grandparent -> Parent -> Child

image.png

```python
class GrandFather:
  def Property(self):
    print("Property in Pune")

class Father(GrandFather):
  def Property1(self):
    print("Property in Delhi")

class Son(Father):
  def Property2(self):
    print("Property in Dubai")

obj=Son()
obj.Property2()   # Child
obj.Property1()   # Father
obj.Property()    # GrandFather
```

5. Hybrid Inheritance: Mixture of 2 or more types of inheritance

image.png

```python
class A:
  def functionA(self):
    print("Function from A")

class B(A):  #B inherits A
  def functionB(self):
    print("Function from B")

class C(B):  # C inherits B
  def functionC(self):
    print("Function from C")

class D(B):  # D Inherits B
  def functionD(self):
    print("Function from D")

class E(B):  # E inherits B
  def functionE(self):
    print("Function from E")

class F(C,D): # F inherits C and D
  def functionF(self):
    print("Function from F")

obj=F()
obj.functionA()
obj.functionB()
obj.functionC()
obj.functionD()
obj.functionF()
```

**2. Method overriding:** Parent and child, have the function with same name and child' functions overrides(overrule) the parent function.


when i create an object of the child class and call the function, the parent function is not going to run, child function is called

```python
class A:  #Parent
  def fun(self):
    print("function A")

class B(A): #Child
  def fun(self):
    print("function B")

obj=B()
obj.fun()
```

**super() function**

When we want to access sometinh of the parent class in the child class, we can use super

super().function_name()

goes to the parents and search for the function

```python
class A:  #Parent
  a=5
  def fun(self):
    print("function A")

class B(A): #Child
  a=10
  def fun(self):
    super().fun()
    print("function B")
    print(self.a)
    print(super().a)

obj=B()
obj.fun()
```

```python
class A:
  def show(self):
    print("A")

class B:
  def show(self):
    print("B")
    super().show()

class C(B,A):
  def show(self):
    print("C")
    super().show()


obj=C()
obj.show()
```

**3. Encapsulation:** It means putting the variables(Properties) and functions(methods) inside a single container called class

We are encapsulating the variables and functions

Encapsulation also provides with access modifiers, which helps us with a certain level of data protection

**Access Modifier**

Tell us who can access and change the data and who can't.

Types:
- **Public:** Anyone can access and modify


- **Protected:** only accessible inside parent class and inside the child class


- **Private:** no one can access and modfied except itself

```python
#public

class A:
  balance=0   #Public data
  def Add_balance(self):
    self.balance+=500

  def check_balance(self):
    print(self.balance)

obj=A()
obj.check_balance()
obj.Add_balance()
obj.check_balance()
print(obj.balance)   # we can access the data
obj.balance=99999999   # we can modify the data
print(obj.balance)
obj.check_balance()
```

```python
# Protected -> not implement property in python
# how to create a protected variable-> put a underscore infront
# variable(public)-----> _variable(protected)

# In python we can access and change the protected variable outside the child and parent class

# Protected in python is used to suggest that this variable is important, please dont try to change it(although we still can do it)
```

```python
class A:
  _var=10   #Protected data

class B(A):
  def show(self):
    print(self._var)   # We can access the protected data inside the child

obj=B()
obj.show()
print(obj._var) # Because Python doesnot support protected we still can access it
obj._var=999999
obj.show()
```

```python
# Private -> can only accessed in the class

# how to create a private variable

# variable(public) ---> _variable(protected) ---> __variable(private)
```

```python
class A:
  __password="Python123"  #Private data

  def show(self):
    print("my password is",self.__password)

obj=A()
obj.show()
obj.__password="Sarthak"   # It is creating a new variable, means that data present inside the class is secure
print(obj.__password)
# Original Value is not affected
obj.show()
```

**4. Abstraction:** It means hidding of unnecessary details and only focusing on important things
- We create some methods(function) which are important in parent class that should be implemented in the child class that we can achieve with the help of abstraction


**Abstract methods:**
- The important method, that should neccessarily be implemented or present in child class

**Abstract class:**
- The parent class which is containing atleast 1 abstract method
- This class give the child class a template

**note:** When we dont implement the important methods in child class, so when i create an object of child class we will not be able to create that object, it will give you an error

image.png

```python
# Bike parent class
# Super bike child class
# without abstraction

class Bike:   #Parent class
  #Important methods
  def start(self):
    pass     # empty Placeholder

  def stop(self):
    pass

  def fuel(self):
    pass

class Super_bike(Bike):
  def start(self):
    print("Super bike has started")

  def fuel(self):
    print("Super bike uses fuel")

  def drift(self):
    print("Super bike can drift")


obj=Super_bike()
obj.start()
```

How to create an abstract class
1. it should have atleast 1 abstract method
2. it should inherit from ABC -> Abstract method class

```python
# abc = Abstract method classes
# ABC = Abstract method class

from abc import ABC, abstractmethod
#                    ^ Create abstract methods

class Bike(ABC):   #Parent class should alway inherit ABC to make it a abstract class

  @abstractmethod  # transform a method into a abstract method
  def start(self):
    pass

  @abstractmethod
  def stop(self):
    pass

  @abstractmethod
  def fuel(self):
    pass

class Super_bike(Bike):
  def start(self):
    print("Super bike has started")

  def fuel(self):
    print("Super bike uses fuel")

  def stop(self):
    print("Super bike has stopped")

  def drift(self):
    print("Super bike can drift")


obj=Super_bike()
obj.start()
```

```python
# RBI -> parent class (Abstract class)
# inside this class you have to create the follow abstract methods
# - Add_money()
# - Deduct_money()
# - Check_balance()

# Child class = SBI_Account, ICICI_Account, ..etc
# each class can include any other method too as per our choice like ATM pin Change, KYC etc

# Time till 10:15 PM
```

```python
from abc import ABC, abstractmethod

class RBI(ABC):   # Abstract class

    @abstractmethod
    def eadd_mony(self, amount):
        pass

    @abstractmethod
    def deduct_money(self, amount):
        pass

    @abstractmethod
    def check_balance(self):
        pass


class SBI_Account(RBI):   #Child class

    def __init__(self):
        self.acc_bal = 1000

    def add_money(self, amount):
        self.acc_bal += amount
        print("Added:", amount)

    def deduct_money(self, amount):
        self.acc_bal -= amount
        print("Deducted:", amount)

    def check_balance(self):
        print("Account Balance:", self.acc_bal)

    def atm_pin(self):
        print("ATM PIN: 1234")


class HDFC_Account(RBI):

    def __init__(self):
        self.acc_bal = 2000

    def add_money(self, amount):
        self.acc_bal += amount
        print("Added:", amount)

    def deduct_money(self, amount):
        self.acc_bal -= amount
        print("Deducted:", amount)

    def check_balance(self):
        print("Account Balance:", self.acc_bal)

    def kyc(self):
        print("KYC completed")


Elon = SBI_Account()
Trump = HDFC_Account()

Elon.add_money(500)
Elon.deduct_money(200)
Elon.check_balance()
Elon.atm_pin()

print()

Trump.add_money(1000)
Trump.deduct_money(500)
Trump.check_balance()
Trump.kyc()
```
