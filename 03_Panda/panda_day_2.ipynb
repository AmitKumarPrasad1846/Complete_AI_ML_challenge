# 🔍 Accessing Elements in a Pandas DataFrame

## 1. Column Selection
* **Single Column (Returns a Series):**
  ```python
  df['Column_Name']   # or df.Column_Name
  ```
* **Multiple Columns (Returns a DataFrame):**
  ```python
  df[['Col1', 'Col2']]
  ```

## 2. Label-Based Indexing (`.loc[]`)
Accesses rows and columns by their **explicit labels/names**.
* **Single Row:** 
  ```python
  df.loc['Row_Label']
  ```
* **Specific Rows & Columns:** 
  ```python
  df.loc[['Row1', 'Row2'], ['Col1', 'Col2']]
  ```
* **Row Slicing:** 
  ```python
  df.loc['Row1':'Row3']   # Note: Includes the endpoint
  ```

## 3. Position-Based Indexing (`.iloc[]`)
Accesses rows and columns by their **integer index positions** (0-indexed).
* **Single Element:** 
  ```python
  df.iloc[row_idx, col_idx]
  ```
* **Row Slicing:** 
  ```python
  df.iloc[0:3]   # Note: Excludes index 3 (standard Python slicing)
  ```
* **Subgrid Selection:** 
  ```python
  df.iloc[0:2, 1:3]
  ```

## 4. Fast Scalar Access (Single Value)
Optimized high-speed lookup for a single cell:
* **By Label:** 
  ```python
  df.at['Row_Label', 'Col_Name']
  ```
* **By Position:** 
  ```python
  df.iat[row_idx, col_idx]
  ```

## 5. Conditional / Boolean Indexing
Filter elements based on logical conditions:
* **Single Condition:** 
  ```python
  df[df['Age'] > 25]
  ```
* **Multiple Conditions:** 
  ```python
  df[(df['Age'] > 25) & (df['Salary'] > 50000)]
  ```
