# 🔍 Exploratory Data Analysis (EDA)

## What is EDA?
Exploratory Data Analysis (EDA) is the essential process of inspecting, summarizing, and understanding your dataset before building or training Machine Learning (ML) models.

## Key Steps in EDA

1. **Load Data:** Import the dataset into Pandas from a source file (CSV, SQL, Excel, etc.).
2. **Understand Structure:** Inspect dimensions (shape), rows, and overall column layout.
3. **Check Data Types:** Verify field types (integers, floats, categorical/objects, datetime).
4. **Find Missing Values:** Scan for null or `NaN` entries across columns.
5. **Handle Missing Values:** Clean data by filling missing entries (imputation) or dropping incomplete rows.
6. **Find Duplicate Values:** Identify repeated identical records in the dataset.
7. **Handle Duplicate Values:** Remove redundant rows to prevent model bias.
8. **Detect Outliers:** Spot extreme anomalies or unusual data points that could distort results.

# 🩹 Filling Missing Values: Mean, Median, and Mode

When handling missing data during EDA, choose the imputation strategy based on data type and distribution:

## Imputation Rules

### 1. Mean
* **When to use:** Numerical data that is **normally distributed** (no significant outliers).
* **Reason:** The average accurately represents the center when there are no extreme values to distort it.

### 2. Median
* **When to use:** Numerical data that contains **outliers** or is skewed.
* **Reason:** The median (middle value) is robust to outliers and prevents extreme data points from distorting the imputed values.

### 3. Mode
* **When to use:** **Categorical** (non-numerical/text) columns.
* **Reason:** Replaces missing entries with the most frequently occurring category.
