```python
import pandas as pd
import numpy as np
```

```python
'''
how pandas works

pandas has 2 structure

1.series -- it is a one-dimensional label array/data
2.dataframe --- it is a two-dimensional label array/data

'''
```

```python
'''
series

*it is 1d labeled array
*it is simple coloumn data with index
*it stores number,string,boolean -- mixed datatype

series=data+index

'''
```

```python
#creating series
s=pd.Series([10,20,30,40,50,60,70,80,90,'a'])
print(s)
print(type(s))
```

```python
#creating series using dict
#key ---> index
#value ---> data
s1=pd.Series({'a':101,'b':'hello','c':'pandas'})
print(s1)
print(type(s1))
```

```python
'''
dataframe

*a dataframe is a 2d labeled,tabular data structure in python
*a table with rows and coloumns
*dataframe == multiple series aligned with index

'''
```

```python
#creating dataframe using dictionary
d1={'a':['raju','amit','smiran'],'b':[101,102,103],'c':['red','yellow','green']}
print(d1)
```

```python
df1=pd.DataFrame(d1)
print(df1)
```

```python
#creating df using list
data=[['rahul',21,86],
      ['priya',22,90],
      ['amit',20,78]]
df=pd.DataFrame(data,columns=['name','age','marks'])
print(df)
```
