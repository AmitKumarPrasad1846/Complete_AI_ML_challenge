# 🐼 Pandas: Python Data Library

## What is Pandas?
Pandas is a powerful Python library specifically designed for **data manipulation** and **data analysis**. 
* **Stands for:** Panel Data System.
* **Creator:** Wes McKinney (2008).

## Why was Pandas Created?
While other libraries handle pure mathematics, Pandas was built to solve **real-world data problems**. It provides tools to easily handle messy data, including:
* Missing values
* Duplicates
* Complex column names and indexing

## Core Capabilities (The Data Pipeline)
Pandas helps you navigate the entire data lifecycle:
1. **Load:** Import data from various sources (CSV, Excel, SQL, etc.).
2. **Clean:** Handle missing data and fix formatting issues.
3. **Manipulate:** Filter, sort, and merge datasets.
4. **Transform & Explore:** Reshape data to find patterns and insights.
5. **Analyze:** Prepare and process data for Machine Learning (ML) models.

```python
import pandas as pd
import numpy as np
```

# 🏗️ How Pandas Works: Core Data Structures

Pandas relies on two primary data structures to organize and manipulate data:

## 1. Series
* **Definition:** A **one-dimensional** labeled array/data.
* **Concept:** Think of it as a single column or a simple list of data where each item is assigned an index (label).

## 2. DataFrame
* **Definition:** A **two-dimensional** labeled array/data.
* **Concept:** Think of it as a table with rows and columns (similar to an Excel spreadsheet or an SQL table). A DataFrame is essentially a collection of Series side-by-side.

# 📊 Pandas Series in Detail

## What is a Series?
* **Structure:** A **one-dimensional** labeled array.
* **Visualization:** Think of it as a simple, single column of data that comes with an attached index (row labels).
* **Flexibility:** It can store mixed data types within the same structure, including:
  * Numbers (integers, floats)
  * Strings (text)
  * Booleans (True/False)

## The Core Concept
At its most fundamental level, a Series is made up of two components:
> **Series = Data + Index**

```python
#creating series
s=pd.Series([10,20,30,40,50,60,70,80,90,'a'])
print(s)
print(type(s))
```

# 📖 Creating a Series Using a Dictionary

You can easily create a Pandas Series using a standard Python dictionary. When you pass a dictionary into a Series, Pandas automatically maps the components as follows:

* **Dictionary Keys** ➡️ Become the Series **Index** (row labels)
* **Dictionary Values** ➡️ Become the Series **Data**

## 💡 Quick Example Concept:

If you have a dictionary like `{'Apples': 5, 'Bananas': 10}`:
* `"Apples"` and `"Bananas"` (the keys) become your **Index**.
* `5` and `10` (the values) become your **Data**.

```python
#creating series using dict
#key ---> index
#value ---> data
s1=pd.Series({'a':101,'b':'hello','c':'pandas'})
print(s1)
print(type(s1))
```

# 🗄️ Pandas DataFrame in Detail

## What is a DataFrame?
* **Structure:** A **two-dimensional**, labeled, tabular data structure in Python.
* **Visualization:** Think of it as a standard data table composed of **rows and columns** (much like an Excel spreadsheet or a SQL table).

## The Core Concept
At its structural level, a DataFrame is simply a collection of Series bound together:
> **DataFrame = Multiple Series aligned by a common Index**

```python
#creating dataframe using dictionary
d1={'a':['raju','amit','smiran'],'b':[101,102,103],'c':['red','yellow','green']}
print(d1)
```

```python
df1=pd.DataFrame(d1)
print(df1)
```

```python
#creating df using list
data=[['rahul',21,86],
      ['priya',22,90],
      ['amit',20,78]]
df=pd.DataFrame(data,columns=['name','age','marks'])
print(df)
```
