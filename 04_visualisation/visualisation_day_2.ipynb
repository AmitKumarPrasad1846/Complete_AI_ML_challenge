# 🥧 Pie Chart in Matplotlib

## What is a Pie Chart?
* **Definition:** A **circular graph** divided into slices (pieces), where each slice represents a proportion of the whole.
* **Core Function:** Used to display **percentage distributions** and relative proportions.

## When to Use?
* Ideal for understanding the **contribution of each category** relative to the total aggregate (parts-of-a-whole relationship).

## How to Plot in Python (`Matplotlib`)
Use the `plt.pie()` function to generate a pie chart:

```python
import matplotlib.pyplot as plt

# 1. Prepare data and category labels
data = [40, 30, 20, 10]
categories = ['Category A', 'Category B', 'Category C', 'Category D']

# 2. Generate the pie chart (autopct displays percentage values)
plt.pie(data, labels=categories, autopct='%1.1f%%', startangle=90)

# 3. Add title and display
plt.title('Contribution of Each Category')
plt.show()
```

````python
# added the AutoPct
# autopct -> to display the percentage of each category

values=[5,20,40,35,18]
categories=['mango','orange','apple','banana','grapes']
colors=['pink','lightgreen','orange','skyblue','red']
plt.pie(values,labels=categories,colors=colors, autopct='%0.4f%%') # autopct
plt.show()
````

````python
# added explode in the pie chart...
#explode ---> to separate the slice from the whole pie chart

values=[5,20,40,35,18]
categories=['mango','orange','apple','banana','grapes']
colors=['pink','lightgreen','orange','skyblue','blue']
plt.pie(values,labels=categories,colors=colors,autopct='%0.2f%%',explode=[0,0.3,0,0.2,0])
plt.show()
````

````python
# Shadow

values=[5,20,40,35,18]
categories=['mango','orange','apple','banana','grapes']
colors=['pink','lightgreen','orange','skyblue','blue']
plt.pie(values,labels=categories,colors=colors,autopct='%0.2f%%',explode=[0,0.3,0,0.2,0], shadow=True)
plt.show()
````

````python
# added wedgeprops - it adds the edge border

values=[5,20,40,35,18]
categories=['mango','orange','apple','banana','grapes']
colors=['pink','lightgreen','orange','skyblue','blue']
plt.pie(values,labels=categories,colors=colors,autopct='%0.2f%%',explode=[0,0.3,0,0.2,0],shadow=True,wedgeprops={'edgecolor':'black'}) 
plt.show()
````

# 📈 Data Distribution Plots

## What is Data Distribution?
* **Definition:** Visualizing how numerical data values are spread out across a range.
* **Purpose:** To see the underlying pattern, central tendency, spread, skewness, probability density, and identify potential **outliers** in a dataset.

## Key Distribution Plots

### 1. Scatter Plot
* **Description:** Displays individual data points plotted along two axes (\(X\) and \(Y\)).
* **When to Use:** To analyze the distribution and relationship/correlation between **two continuous variables**.
* **How to Plot:** `plt.scatter(x, y)` or `sns.scatterplot(x=x, y=y)`

### 2. Box Plot (Box-and-Whisker Plot)
* **Description:** Summarizes data distribution using a **5-number summary** (Minimum, \(Q_1\), Median, \(Q_3\), Maximum).
* **When to Use:** To compare distributions across groups, observe data symmetry/spread, and spot **outliers**.
* **How to Plot:** `plt.boxplot(data)` or `sns.boxplot(data=data)`

### 3. Violin Plot
* **Description:** A hybrid visual that combines a **Box Plot** with a **Kernel Density Plot (KDE)**.
* **When to Use:** To see the exact shape of the distribution (e.g., single peak vs. double peaks) alongside summary statistics.
* **How to Plot:** `sns.violinplot(data=data)`

## How to Plot in Python (`Matplotlib` / `Seaborn`)

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Sample Data Execution

# 1. Scatter Plot
plt.scatter(x_data, y_data, color='blue')
plt.title('Scatter Plot')
plt.show()

# 2. Box Plot
sns.boxplot(y=data, color='lightgreen')
plt.title('Box Plot')
plt.show()

# 3. Violin Plot
sns.violinplot(y=data, color='purple')
plt.title('Violin Plot')
plt.show()
```

# 📍 Scatter Plot in Matplotlib

## What is a Scatter Plot?
* **Definition:** A type of plot that displays individual data points on a **2-dimensional (2D) plane**.
* **Visual Representation:** Each dot represents an individual data point plotted using two values as coordinates \((X, Y)\).

## When to Use?
* **Show Relationships:** Analyze the correlation or relationship between **two numerical variables** (e.g., height vs. weight, study hours vs. test scores).
* **Visualize Data Spread:** Observe overall patterns, clusters, trends, and the dispersion of data points.

## How to Plot in Python (`Matplotlib`)
Use the `plt.scatter()` function to generate a scatter plot:

```python
import matplotlib.pyplot as plt

# 1. Prepare two numerical datasets
x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78]

# 2. Generate the scatter plot
plt.scatter(x, y, color='coral', marker='o')

# 3. Add labels and display
plt.xlabel('Variable X (Numerical)')
plt.ylabel('Variable Y (Numerical)')
plt.title('Scatter Plot Example')
plt.show()
```

````python
rollno=[1,2,3,4,5,6,7,8,9]
marks=[99,22,45,62,67,40,82,85,80]
plt.scatter(rollno,marks)
plt.show()
````

````python
#making scatter plot more interactive

rollno=[1,2,3,4,5,6,7,8,9]
marks=[99,22,45,62,67,40,82,85,80]
plt.scatter(rollno,marks,color='red',marker='*')
plt.title('rollno vs marks')
plt.xlabel('rollno')
plt.ylabel('marks')
plt.grid()
plt.show()
````

````python
# added the marker size...

rollno=[1,2,3,4,5,6,7,8,9]
marks=[99,22,45,62,67,40,82,85,80]
plt.scatter(rollno,marks,color='red',marker='s',s=150)
plt.title('rollno vs marks')
plt.xlabel('rollno')
plt.ylabel('marks')
plt.grid() #(axis='y'--> to get only horizontal line)
plt.show()
````

````python
#multiple datapoint

rollno=[1,2,3,4,5]
englis_marks=[99,22,45,62,67]
maths_marks=[39,55,61,89,70]
plt.scatter(rollno,englis_marks,color='red',marker='*',s=150,label='englishmarks')
plt.scatter(rollno,maths_marks,color='yellow',marker='s',s=100,label='mathsmarks')
plt.title('rollno vs marks')
plt.xlabel('rollno')
plt.ylabel('marks')
plt.legend(loc='lower right')
plt.grid() 
plt.show()
````

# Threshold line
- it is a refrence line drawn on scatter plot
 1. `plt.axhline` ---> (horizontal line)
 2. `plt.axvline` ---> (vertical line)

````python
rollno=[1,2,3,4,5,6,7,8,9]
marks=[99,22,45,67,88,90,40,82,89]
plt.scatter(rollno,marks,color='red',marker='*',s=150)
plt.title('rollno vs marks')
plt.xlabel('rollno')
plt.ylabel('marks')
plt.axhline(y=50,color='blue')
plt.axvline(x=5,color='red',linestyle='--')
plt.show()
````
