# What is visualisation ?
- Data visualisation is the graphical representation of the data using charts, graphs, plots, so that humans can easily understand the patterns, trends, insights.

# Why we need visualisation ?
- To understand larger amount of Data.
- To find the patterns, trends and insights.
- To make decisions.
- To compare the values.
