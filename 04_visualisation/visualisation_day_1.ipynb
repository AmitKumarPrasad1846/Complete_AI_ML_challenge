# 1. Line plot
- a line plot is a type of graph that dispay the datapoint connected by straight line

- **when to use** --> `continuous data,to see the trend`
- **how to plot** --> `plt.plot()`

````python
days = ['mon','tue','wed','thur','fri','sat','sun']
sales = [480,500,250,190,382,650,700]
plt.plot(days,sales) # plt.plot = it is used for the "line plot"
plt.show()
````

````python
# making the representation more colorful and adding title and label.

days = ['mon','tue','wed','thur','fri','sat','sun']
sales = [480,500,250,190,382,650,700]
plt.plot(days,sales, color='red')
plt.title("sales vs days graph")
plt.xlabel('days', color='green')
plt.ylabel('sales', color='green')
plt.show()
````

````python
days = ['mon','tue','wed','thur','fri','sat','sun']
sales = [480,500,250,190,382,650,700]
plt.plot(days,sales, color='red', marker='s')

# check the marker_codes.png for more markers...

plt.title("sales vs days graph")
plt.xlabel('days', color='green')
plt.ylabel('sales', color='green')
plt.show()
````

````python
# added "markerfacecolor" and "linestyle" in plt.plot

days = ['mon','tue','wed','thur','fri','sat','sun']
sales = [480,500,250,190,382,650,700]
plt.plot(days,sales, color='red', marker='s', markerfacecolor='yellow', linestyle='-.')
plt.title("sales vs days graph")
plt.xlabel('days', color='green')
plt.ylabel('sales', color='green')
plt.show()
````

* '- ' ---> solid line
* '--' --> dashed line
* '-.' ---> dash-dot
* ':' --> dotted line

````python
# added 'markeredgecolor' in the plt.plot

days = ['mon','tue','wed','thur','fri','sat','sun']
sales = [480,500,250,190,382,650,700]
plt.plot(days,sales, color='orange', marker='s', markerfacecolor='purple', linestyle='-.', markeredgecolor='black')
plt.title("sales vs days graph")
plt.xlabel('days', color='pink')
plt.ylabel('sales', color='pink')
plt.show()
````

````python
# added "plt.grid()"

days = ['mon','tue','wed','thur','fri','sat','sun']
sales = [480,500,250,190,382,650,700]
plt.plot(days,sales, color='orange', marker='s', markerfacecolor='purple', linestyle='-.', markeredgecolor='black')
plt.title("sales vs days graph")
plt.xlabel('days', color='pink')
plt.ylabel('sales', color='pink')
plt.grid() # it will add the grid on the plot
plt.show()
````

````python
# multiple line in a single plot...

days=['mon','tue','wed','thur','fri','sat','sun']
mango_sales=[450,200,380,420,550,590,390]
banana_sales=[100,280,380,400,780,820,760]
apple_sales=[340,500,620,800,120,400,280]

plt.plot(days,mango_sales,color='orange',linestyle='-',marker='o',markerfacecolor='blue',label='mangosales')
plt.plot(days,banana_sales,color='yellow',linestyle='--',marker='s',markerfacecolor='red',label='bananasales')
plt.plot(days,apple_sales,color='red',linestyle='-.',marker='*',markerfacecolor='green',label='applesales')

plt.grid() # it will add grid in the plot
plt.show() # it will print the plot
````

````python
days=['mon','tue','wed','thur','fri','sat','sun']
mango_sales=[450,200,380,420,550,590,390]
banana_sales=[100,280,380,400,780,820,760]
apple_sales=[340,500,620,800,120,400,280]

plt.plot(days,mango_sales,color='orange',linestyle='-',marker='o',markerfacecolor='blue',label='mangosales')
plt.plot(days,banana_sales,color='yellow',linestyle='--',marker='s',markerfacecolor='red',label='bananasales')
plt.plot(days,apple_sales,color='red',linestyle='-.',marker='*',markerfacecolor='green',label='applesales')

plt.grid() # it will add grid in the plot
plt.legend() # tells more about the marker
plt.show() # it will print the plot
````
