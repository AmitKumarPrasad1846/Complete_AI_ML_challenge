# 📊 Histograms in Data Visualization

## What is a Histogram?
* **Definition:** A graph used to show the **distribution of continuous numerical data**.
* **How it Works:** It groups data into range intervals called **bins** and counts how many values fall within each range.

## Example Concept
Grouping student exam scores into bins:
* **Score Range (10–30):** 3 students
* **Score Range (31–50):** 12 students
* **Score Range (51–70):** 8 students

## Syntax (using Matplotlib)
```python
import matplotlib.pyplot as plt

plt.hist(data, bins=10)  # Generates the histogram plot
plt.show()

````python
# Interactive Histogram.

# edgecolor = give color to the bin image.
# bins = regtangular bars or the no. of intervals(groups).

age=[34, 45, 10, 15, 13, 18, 19, 55, 28, 29, 25, 33, 30, 50, 48, 49, 33, 30, 17]
plt.hist(age,color='pink',bins=8,edgecolor='black')
plt.show()
````

````python
# range = it is the range of data in which the graph wil get plotted...

age=[34, 45, 10, 15, 13, 18, 19, 55, 28, 29, 25, 33, 30, 50, 48, 49, 33, 30, 17]
plt.hist(age,color='orange',bins=6,edgecolor='black', range=(10,60))
plt.show()
````
