# 📊 Histograms in Data Visualization

## What is a Histogram?
* **Definition:** A graph used to show the **distribution of continuous numerical data**.
* **How it Works:** It groups data into range intervals called **bins** and counts how many values fall within each range.

## Example Concept
Grouping student exam scores into bins:
* **Score Range (10–30):** 3 students
* **Score Range (31–50):** 12 students
* **Score Range (51–70):** 8 students

## Syntax (using Matplotlib)
```python
import matplotlib.pyplot as plt

plt.hist(data, bins=10)  # Generates the histogram plot
plt.show()

# 🎻 Violin Plot

## What is a Violin Plot?
* **Definition:** A graph used to show the **distribution of continuous numerical data across different categories**.
* **Hybrid Structure:** It combines a **Box Plot** with a **KDE (Kernel Density Estimate) / Density Plot**.

## Key Features & How to Read
* **Density Representation:** It displays the probability density of the data at different values.
* **Shape Meaning:** 
  * **Thick sections:** Indicate a high concentration/density of data points at that value.
  * **Thin sections:** Indicate fewer data points at that value.

## Syntax (using Matplotlib)
```python
import matplotlib.pyplot as plt

plt.violinplot(data)  # Generates the violin plot
plt.show()

````python
marks=[40,45,50,52,55,60,65,67,88,90,72,66,70,75,80,85]
plt.violinplot(marks)
plt.show()
````

````python
# showmeans = True - it shows the mean of the data...
# widths - it controls the width of the violin

marks=[40,45,50,52,55,60,65,67,88,90,72,66,70,75,80,85]
plt.violinplot(marks, showmeans=True, widths=0.8)
plt.show()
````

````python
marks=[40,45,50,52,55,60,65,67,88,90,72,66,70,75,80,85,42,45,43,46,47,48,49,41]
plt.violinplot(marks,showmeans=True,widths=0.8)
plt.show()
````
